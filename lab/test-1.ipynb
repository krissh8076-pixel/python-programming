```python
def moving_average(stream, w):
    result = []
    current_sum = 0
    for i in range(len(stream)):
        current_sum += stream[i]
        if i >= w:
            current_sum -= stream[i - w]
        window_size = min(i + 1, w)

        result.append(current_sum / window_size)

    return result
stream = [1, 2, 3, 4, 5]
w = 3
print(moving_average(stream, w))
```
