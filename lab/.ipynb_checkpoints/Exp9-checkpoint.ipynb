### Ques1: Student class

```python
class Student:
    def __init__(self, name, sap, phy, chem, maths):
        self.name = name 
        self.sap = sap
        self.marks = [phy, chem, maths]

    def display(self):
        print(self.name, self.sap, self.marks)

    def percentage(self):
        return sum(self.marks)/3

    def result(self):
        if all(m > 40 for m in self.marks):
            return "Pass"
        return "Fail"
    
```

### Ques4: Operator Overloading

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Point(self.x + other.x, self.y +other.y)

    def display(self):
        print(self.x, self.y)

p1 = Point(20,30)
p2 = Point(14,15)
p3 = p1 + p2
p3.display()
```
