### Ques1: Names in file 

```python
with open("/Users/krishsinghal/Desktop/python-programming-1/lab/name.txt", "r") as f:
    names = [name.strip() for name in f if name.strip()]

print("Total names:", len(names))

vowels = "AEIOUaeiou"
count_vowel = sum(1 for name in names if name[0] in vowels)

print("Names starting with vowel:", count_vowel)

longest = max(names, key=len)
print("Longest name:", longest)
```

### Ques2: Integers in file 

```python
with open("/Users/krishsinghal/Desktop/python-programming-1/lab/numbers.txt", "r") as f:
    nums = [int(x.strip()) for x in f if x.strip()]

print("Max:", max(nums))
print("Average:", sum(nums)/len(nums))
print("Count >100:", len([x for x in nums if x > 100]))
```

### Ques3: City File Handling

```python
with open("/Users/krishsinghal/Desktop/python-programming-1/lab/city.txt", "r") as f:
    cities = [line.split() for line in f]

print("All Cities:")
for city in cities:
    print(city)

print("\nCities with population > 10 lakhs:")
for city in cities:
    if float(city[1]) > 10:
        print(city[0])

total_area = sum(float(city[2]) for city in cities)
print("\nTotal Area:", total_area)
```

### Ques4: Exception Handling

```python
n = int(input())

for _ in range(n):
    try:
        a, b = input().split()
        print(int(a) // int(b))
    except ZeroDivisionError as e:
        print("Error Code:", e)
    except ValueError as e:
        print("Error Code:", e)
```

### Ques5: Create multiple suitable exceptions for a file handling program

```python
try:
    filename = input("Enter file name: ")
    
    f = open(filename, "r")
    
    content = f.read()
    
    if content == "":
        raise Exception("File is empty")
    
    print("File Content:\n", content)
    
    f.close()

except FileNotFoundError:
    print("Error: File does not exist.")

except PermissionError:
    print("Error: You don’t have permission to access this file.")

except IsADirectoryError:
    print("Error: You entered a directory instead of a file.")

except Exception as e:
    print("General Error:", e)

finally:
    print("Execution completed.")
```
