### Ques1: Find Max and Min without built-in functions 

```python
def find_max_min(lst):
    maximum = lst[0]
    minimum = lst[0]

    for num in lst:
        if num > maximum:
            maximum = num
        if num > minimum:
            minimum = num

    return maximum, minimum 
```

### Ques2: Sum of cubes less than n

```python
def sum_of_cubes(n):
    total = 0
    for i in range(1,n):
        total += i**3
    return total 
```

### Ques3: Print 1 to n using recursion(no loop)

```python
def print_n(n):
    if n == 0:
        return
    print_n(n-1)
    print(n)
```

### Ques4: Recursive Fibonacci up to n terms

```python
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n-1) + fibonacci(n-2)

def print_fibonacci(n):
    for i in range(n):
        print(fibonacci(i), end=" ")
```

 ### Ques5: Lambda Function for volume of cone

```python
volume_cone = lambda r, h: (3.14 * r * r * h) / 3
```

***Explanation:***

### Ques6: Lamda returning max and min tuple

```python
max_min = lambda lst: (max(lst), min(lst))
```

***Explanation:***

- Compact one line function

### Ques7: Argument Types: a) Keyword b) Default c) Variable length

```python
def greet(name, age):
    print(name, age)

greet(age=19, name="Krrish")
```

```python
def greet(name="Guest"):
    print("Hello", name)
greet()
```

```python
def add(*numbers):
    return sum(numbers)
    
```

***Explanation:***
- *numbers accepts multiple values
- arguments are satored as a tuple
- works for any no of inputs
- very flexible design
