# Day 5 – While Loop

# 1. Reverse Number

```python
n = 123
rev = 0
while n > 0:
    rev = rev*10 + n%10
    n //= 10
print(rev)
```

# 2. Palindrome Number

```python
n = 121
temp = n
rev = 0
while n > 0:
    rev = rev*10 + n%10
    n //= 10
print("Palindrome" if temp == rev else "Not")
```

# 3. Fibonacci

```python
a, b = 0, 1
for _ in range(10):
    print(a)
    a, b = b, a+b
```

# 4. Count Digits

```python
n = 12345
count = 0
while n > 0:
    count += 1
    n //= 10
print(count)
```

# Day 10 – Functions

```python
# 1
def greet():
    print("Hello")
greet()
```

```python
# 2
def add(a,b):
    return a+b
print(add(2,3))
```

```python
# 3
def fact(n):
    return 1 if n==0 else n*fact(n-1)
print(fact(5))
```
